```python
# Lendo os dados vindos de uma planilha do Google Sheets usando Python

import pandas as pd

url_planilha_veiculos = "https://docs.google.com/spreadsheets/d/e/2PACX-1vRYEMeo5x9R-JHkn8QSVYlwAswV5sZOVLjpt5MLwue_ag5V728gHzWzAfPNPQx4Y4kbMwKChiJIfHsw/pub?gid=483992714&single=true&output=csv"

df_veiculos = pd.read_csv(url_planilha_veiculos)

display(df_veiculos)
```

```python
# criar um script em python que faça o usuario adicionar novos dados na planilha
import requests
import json

endpoint_sheets = "https://script.google.com/macros/s/AKfycbzpSoGJfY3I3_fb32OeaL9ym-3roE2x-r8vCbpQ2PT3OVkfX29Mz9IPVh2BZMEjm_MrKw/exec"
def enviar_msg(id_veiculo, nome_veiculo, operador, data_aquisicao):
  response = requests.post(
      endpoint_sheets,
      json={'id_veiculo':id_veiculo,'nome_veiculo':nome_veiculo,'operador':operador, 'data_aquisicao':data_aquisicao }
  )
  print(response.text)
```

```python
id_veiculo = input("informe o ID do seu veículo: ")
nome_veiculo = input("informe o nome do seu veículo: ")
operador = input("informe o operador do veículo: ")
data_aquisicao = input("informe a data da aquisição do seu veículo: ")

enviar_msg(id_veiculo, nome_veiculo, operador, data_aquisicao)
```
