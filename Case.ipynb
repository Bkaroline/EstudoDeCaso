```python
import pandas as pd

# --- INGESTÃO E LEITURA
dados_entrega = "/content/entregas_orquideas.csv"
dados_qualidade = "/content/estufas_qualidade.csv"
dados_telemetria = "/content/telemetria_iot.csv"

df_dados_entrega = pd.read_csv(dados_entrega)
df_dados_qualidade = pd.read_csv(dados_qualidade)
df_dados_telemetria = pd.read_csv(dados_telemetria)

display(df_dados_entrega.head())
print("\n\n")
display(df_dados_qualidade.head())
print("\n\n")
display(df_dados_telemetria.head())
print("\n\n")

# --- CLASSIFICAÇÃO TEMPERATURA

status_temperatura = []

Temp = df_dados_telemetria["Temperatura_C"].tolist()
for i in Temp:
  if i >= 28:
    status_temperatura.append("ALERTA")
  else:
    status_temperatura.append("OK")


# --- ANALISES DE ESTATISTICA E DIAGNOSTICO

df_dados_telemetria["Classificacao_Temp"] = status_temperatura
display(df_dados_telemetria.head())
print("\n\n")

print(f"O total de Orquideas transportadas foram: {df_dados_entrega["Qtd_Orquideas"].sum()}")
print("\n\n")

df_preju = df_dados_entrega[df_dados_entrega["Status_Entrega"]== "Cancelado"]
print(f"A Quantidade de Orquideas canceladas foram: {df_preju["Status_Entrega"].count()}")
print("\n\n")

df_est = df_dados_entrega[df_dados_entrega["Status_Entrega"]== "Cancelado"]
print(f"O estado que mais apareceu foi: {df_est["UF_Destino"].max()}")
print("\n\n")

print(f"A Temperatura máxima foi:  {df_dados_telemetria["Temperatura_C"].max()}°C e a temperatura média foi: {df_dados_telemetria["Temperatura_C"].mean()}°C")
print("\n\n")
```
